```python
a = "Life is too short, you need Python"
print(len(a))

print(a[3])     # e
print(a[12])    # s
print(a[-1])    # n
print(a[-5])    # y

print("")

# Python 레인지 확인
print(a[0]+a[1]+a[2]+a[3])
print(a[0:4])               # 0 ~ 3 까지
print(a[0:3])               # 0 ~ 2 까지
print(a[5:7])               # 5 ~ 6 까지
print(a[5:6])
print(a[0]+a[1]+a[3])
```

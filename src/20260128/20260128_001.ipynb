# Tkinder (Tk interface)

* Python에 기본 포함된 GUI(그래픽 사용자 인터페이스) 라이브러리로 버튼, 텍스트 상자, 라벨 등 윈도우 창을 만들고 제어할 수 있는 기능
* Tk: 1991년에 개발된 Tcl(Tool Command Language) 언어용 GUI 툴킷

```python
from tkinter import *    # tkinter 모듈의 모든(*) 클래스와 함수 불러오기

root = Tk()     ## 메인 윈도우 창 객체 생성
root.geometry("300x200")  ## 창 크기(가로X세로)

label = Label(root, text="Hello, World!")    ## 텍스트를 표시하는 라벨(Label) 위젯
label.pack()     ## 위에서 만든 라벨을 창에 실제로 배치

root.mainloop()   ## 반드시 입력해야만, 우리 눈에 창이 보임, 종료버튼을 클릭할때까지 창이 열려있게 함.
```

```python
from tkinter import *

def on_enter(e):
    # 마우스 오버 시: 주황색 배경, 흰색 글자
    button.config(bg="green", fg="white")

def on_leave(e):
    # 마우스가 떠날 때: 리눅스/윈도우 공용 기본 색상인 'lightgray' 등으로 설정
    # 또는 원래 기본값인 button.default_bg 사용 (아래 참조)
    button.config(bg="lightgrey", fg="black")

root = Tk()
root.geometry("300x200")

# 버튼 생성 시 기본 배경색을 명시적으로 지정하는 것이 안전합니다.
button = Button(root, text="마우스를 올려보세요", width=20, height=2, bg="blue", fg="white")
button.pack(pady=50)

button.bind("<Enter>", on_enter)
button.bind("<Leave>", on_leave)

root.mainloop()
```

```python
from tkinter import *

def submit(event=None):
    name = entry_name.get()
    age = entry_age.get()
    print("이름:", name)
    print("나이:", age)

root = Tk()
root.geometry("300x200")

Label(root, text="이름:").pack()  

entry_name = Entry(root)
entry_name.pack()

Label(root, text="나이:").pack()

entry_age = Entry(root)
entry_age.pack()

button_submit = Button(root, text="제출", command=submit)
button_submit.pack()

root.bind('<Return>', submit)

root.mainloop()
```

```python
## 입력 데이터를 csv로 저장

from tkinter import *
import csv
import os

def submit():
    name = entry_name.get()
    age = entry_age.get()

    file_exists = os.path.exists("../../data/tkinter_data001.csv")
    with open("../../data/tkinter_data001.csv", "a", newline="", encoding="utf-8-sig") as f:
        writer = csv.writer(f)
        if not file_exists:
            writer.writerow(["이름", "나이"])
        writer.writerow([name, age])

    print("이름:", name)
    print("나이:", age)

    # 입력창 초기화
    entry_name.delete(0, END)
    entry_age.delete(0, END)

root = Tk()
root.geometry("300x200")

Label(root, text="이름:").pack()
entry_name = Entry(root)
entry_name.pack()

Label(root, text="나이:").pack()
entry_age = Entry(root)
entry_age.pack()

button_submit = Button(root, text="제출", command=submit)
button_submit.pack()

root.mainloop()  
```

```python
from tkinter import *
from tkinter import messagebox
import csv
import os

# --- CSV 파일 경로 (폴더 없이 현재 위치) ---
file_path = "../../data/tkinter_customers.csv"  # 폴더 없이 파일 저장

# --- CSV 헤더 ---
headers = ["이름", "주소", "전화번호", "나이", "성별"]

# --- 제출 버튼 함수 ---
def submit():
    name = entry_name.get().strip()
    address = entry_address.get().strip()
    phone = entry_phone.get().strip()
    age = entry_age.get().strip()
    gender = gender_var.get()

    if not name:
        messagebox.showwarning("경고", "이름을 입력하세요")
        return

    file_exists = os.path.exists(file_path)   # 파일 존재 여부 확인

    with open(file_path, "a", newline="", encoding="utf-8-sig") as f:
        writer = csv.writer(f)
        if not file_exists:  # 파일이 없으면 항목이름 작성
            writer.writerow(headers)
        writer.writerow([name, address, phone, age, gender])  # 파일 있으면 입력한 내용만 추가

    messagebox.showinfo("완료", f"{name}님 정보가 저장되었습니다")
    clear_entries()

# --- 수정 버튼 함수 ---
def modify():
    name = entry_name.get().strip()
    if not name:
        messagebox.showwarning("경고", "수정할 이름을 입력하세요")
        return

    if not os.path.exists(file_path):
        messagebox.showerror("오류", "저장된 데이터가 없습니다")
        return

    # CSV 읽어서 수정
    updated = False
    rows = []
    with open(file_path, "r", newline="", encoding="utf-8-sig") as f:
        reader = csv.DictReader(f)
        for row in reader:
            if row["이름"] == name:  # 입력한 이름을 찾아서
                row["주소"] = entry_address.get().strip()    # 입력된 데이터로 수정
                row["전화번호"] = entry_phone.get().strip()  # 입력된 데이터로 수정
                row["나이"] = entry_age.get().strip()        # 입력된 데이터로 수정
                row["성별"] = gender_var.get()               # 입력된 데이터로 수정
                updated = True
            rows.append(row)

    if updated:
        # CSV 덮어쓰기
        with open(file_path, "w", newline="", encoding="utf-8-sig") as f:
            writer = csv.DictWriter(f, fieldnames=headers)
            writer.writeheader()
            writer.writerows(rows)
        messagebox.showinfo("완료", f"{name}님 정보가 수정되었습니다")
        clear_entries()
    else:
        messagebox.showwarning("경고", f"{name}님 정보를 찾을 수 없습니다")

# --- 입력창 초기화 함수 ---
def clear_entries():
    entry_name.delete(0, END)
    entry_address.delete(0, END)
    entry_phone.delete(0, END)
    entry_age.delete(0, END)
    gender_var.set("남")

# --- GUI ---
root = Tk()
root.title("고객 정보 관리")
root.geometry("400x350")

# 이름
Label(root, text="이름:").pack(anchor=W, padx=10)
entry_name = Entry(root)
entry_name.pack(fill=X, padx=10, pady=2)

# 주소
Label(root, text="주소:").pack(anchor=W, padx=10)
entry_address = Entry(root)
entry_address.pack(fill=X, padx=10, pady=2)

# 전화번호
Label(root, text="전화번호:").pack(anchor=W, padx=10)
entry_phone = Entry(root)
entry_phone.pack(fill=X, padx=10, pady=2)

# 나이
Label(root, text="나이:").pack(anchor=W, padx=10)
entry_age = Entry(root)
entry_age.pack(fill=X, padx=10, pady=2)

# 성별
Label(root, text="성별:").pack(anchor=W, padx=10)
gender_var = StringVar(value="남")
Frame_gender = Frame(root)
Frame_gender.pack(anchor=W, padx=10, pady=2)
Radiobutton(Frame_gender, text="남", variable=gender_var, value="남").pack(side=LEFT)
Radiobutton(Frame_gender, text="여", variable=gender_var, value="여").pack(side=LEFT)

# 버튼
Frame_buttons = Frame(root)
Frame_buttons.pack(pady=15)
Button(Frame_buttons, text="제출", width=10, command=submit).pack(side=LEFT, padx=5)
Button(Frame_buttons, text="수정", width=10, command=modify).pack(side=LEFT, padx=5)

root.mainloop()
```

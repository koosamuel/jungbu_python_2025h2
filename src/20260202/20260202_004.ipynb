```python
from deep_translator import GoogleTranslator
import tkinter as tk
from tkinter import ttk, messagebox

# 번역기 객체
translator = GoogleTranslator(source='ko', target='en')

def translate():
    try:
        input_text = input_box.get("1.0", tk.END).strip()
        if not input_text:
            messagebox.showwarning("경고", "번역할 텍스트를 입력하세요.")
            return

        target_lang = lang_var.get()
        translator.target = target_lang

        translated = translator.translate(input_text)

        output_box.config(state='normal')
        output_box.delete("1.0", tk.END)
        output_box.insert(tk.END, translated)
        output_box.config(state='disabled')

    except Exception as e:
        messagebox.showerror("오류", f"번역 중 오류 발생: {str(e)}")

# ----------------- GUI -----------------
root = tk.Tk()
root.title("실시간 한글 번역기")
root.geometry("500x300")

tk.Label(root, text="한글 입력:").pack(pady=5)
input_box = tk.Text(root, height=3, width=50)
input_box.pack(pady=5)

tk.Label(root, text="대상 언어:").pack(pady=5)
lang_var = tk.StringVar(value='en')
lang_options = ['en', 'es', 'fr', 'ja', 'zh-CN', 'de', 'ko']
lang_menu = ttk.Combobox(root, textvariable=lang_var, values=lang_options)
lang_menu.pack(pady=5)

tk.Button(root, text="번역", command=translate).pack(pady=10)

tk.Label(root, text="번역 결과:").pack(pady=5)
output_box = tk.Text(root, height=3, width=50, state='disabled')
output_box.pack(pady=5)

root.mainloop()
```

```python
# 경고 메시지 무시
import warnings
warnings.filterwarnings(action='ignore') 

import pandas as pd
from openpyxl import load_workbook
```

```python
info = pd.read_excel("../../data/직원 정보.xlsx", sheet_name = "Sheet1")
info
```

## 텍스트 함수

```python
info["성명"] = info["성"] + info["이름"]   
info["성명1"] = info[["성", "이름"]].sum(axis=1)
info
```

```python
info["사원번호 앞 4자리"] = info["사원번호"].str[0:4]
info["전화번호 뒤 4자리"] = info["전화번호"].str[9:13] # 전화번호 뒤에서 4자리 추출
# info["전화번호 뒤 4자리"] = info["전화번호"].str[-4:]
info
```

```python
info["구"] = info["주소"].str.split(" ").str[1]
info[["성", "이름", "전화번호", "구"]]
```

```python
info["대문자"] = info["영문명"].str.upper()      # [영문명] 열을 대문자로 변환
info["첫 글자"] = info["영문명"].str.capitalize()   # [영문명] 열의 첫 글자만 대문자로 변환
info["단어 첫 글자"] = info["영문명"].str.title()    # 단어 첫글자만 대문자로
info[["순번", "성", "이름", "영문명", "전화번호", "대문자", "첫 글자","단어 첫 글자"]]
```

```python
str.swapcase("ABcde")
```

```python
info["phone"] = info["전화번호"].str.replace("-", "", 1)    # 첫 번째 하이픈(-) 제거 
info["phone1"] = info["전화번호"].str.replace("-"," ")      # 모든 하이픈 (-)을 공백으로 대체
info[["순번", "성", "이름", "전화번호", "phone", "phone1"]]
```

```python
# 문자열 길이
info["주소 길이"] = info["주소"].str.len()   
info[["순번", "성", "이름", "주소", "주소 길이" ]]
```

```python
info["앞뒤 공백 제거"] = info["주소"].str.strip()   
info["공백 제거 후 길이"] = info["앞뒤 공백 제거"].str.len()   # 공백 제거 문자열의 길이 구하기
info[["순번", "주소", "주소 길이", "앞뒤 공백 제거", "공백 제거 후 길이"]]
```

```python
info.head()
```

```python
# 새로운 시트로 추가
with pd.ExcelWriter("../../data/직원 정보.xlsx", engine='openpyxl', mode='a') as writer:
    info.to_excel(writer, sheet_name="문자열함수", index=False)
```

## 수학 및 통계 함수

```python
score = pd.read_excel("../../data/성적 처리.xlsx", sheet_name = "Sheet1")
score
```

```python
score["sum"] = score.iloc[:, 2:7].sum(axis=1)
score
```

```python
score["mean"] = score.iloc[:, 2:7].mean(axis=1)
score
```

```python
score1 = score[["반","국어","영어","수학","사회","과학"]].groupby("반").sum()
score1
```

```python
score2 = score[["반","국어","영어","수학","사회","과학"]].groupby("반").mean()
score2
```

```python
score["순위"] = score["mean"].rank(ascending = False).astype(int)
score
```

```python
score["순위2"] = score["mean"].rank(method="dense", ascending = False).astype(int)
score

# method="max"   :  3위가 2개일때, 4위로 표시하고, 다음 순위는 5위로 
# method="min"   :  3위가 2개일때, 3위로 표시하고, 다음 순위는 5위로 
# method="dense" :  3위가 2개일때, 3위로 표시하고, 다음 순위는 4위로
```

```python
# 새로운 시트로 엑셀파일에 추가
with pd.ExcelWriter("../../data/성적 처리.xlsx", engine='openpyxl', mode='a') as writer:
    score.to_excel(writer, sheet_name="수학통계함수", index=False)
```

## 식자재 주문

```python
product = pd.read_excel("../../data/식자재 주문.xlsx", sheet_name = "product")
order = pd.read_excel("../../data/식자재 주문.xlsx", sheet_name = "order")
display(product, order)
```

```python
mapping = {"1":"청주", "2":"대구", "3":"광주"}

product["공장"] = product["제품코드"].str[-1].map(mapping)
product.head()
```

```python
with pd.ExcelWriter("../../data/식자재 주문.xlsx", engine='openpyxl', mode='a') as writer:
    product.to_excel(writer, sheet_name='공장', index=False)
```

```python
product = pd.read_excel("../../data/식자재 주문.xlsx", sheet_name = "product")
order = pd.read_excel("../../data/식자재 주문.xlsx", sheet_name = "order" )
display(product, order)
```

```python
product.set_index("제품코드", inplace = True)	# product["제품코드"]를 인덱스로 설정
order.set_index("제품코드", inplace = True)		# order["제품코드"]를 인덱스로 설정
```

```python
display(product, order)
```

```python
order["제품명"] = product["제품명"]	 ## 동일한 인덱스의 product["제품명"]을 order에 추가
order["단가"] = product["단가"]	
order["주문금액"] = order["주문수량"] * order["단가"]

display(product, order)
```

```python
order.reset_index(inplace = True)
order
```

```python
with pd.ExcelWriter("../../data/식자재 주문.xlsx", engine='openpyxl', mode='a') as writer:
    order.to_excel(writer, sheet_name='주문', index=False)
```

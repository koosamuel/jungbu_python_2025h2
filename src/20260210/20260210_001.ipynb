# CNN 딥러닝 _ (Convolutional Neural Network, 합성곱 신경망)

* 이미지 처리와 비전 관련 작업에서 뛰어난 성능을 보이는 딥러닝 모델
* 이미지 분류, 객체 탐지, 얼굴 인식, 영상 분석 등 다양한 분야에서 활용
* CNN은 합성곱(convolution) 연산을 사용하여 데이터의 특징을 추출하고 이를 기반으로 분류, 예측 등의 작업을 수행

<br><br>

<br><br>


```python
# 경고 메시지 무시
import warnings
warnings.filterwarnings(action='ignore')

import tensorflow as tf
from keras.models import Sequential
from keras.layers import Conv2D, MaxPooling2D, Flatten, Dense
```

```python
model = Sequential()   # 레이어를 순서대로 쌓는 신경망 모델("순서대로 쌓는 상자"를 준비)

# 첫 번째 합성곱층 (Convolutional Layer)
model.add(Conv2D(32, (3, 3), activation='relu', input_shape=(64, 64, 3)))   # 1(흑백), 3(컬러), 4(컬러+투명도)

# 풀링층 (Pooling Layer)
model.add(MaxPooling2D(pool_size=(2, 2)))

# 두 번째 합성곱층 (Convolutional Layer)
model.add(Conv2D(64, (3, 3), activation='relu'))

# 두 번째 풀링층 (Pooling Layer)
model.add(MaxPooling2D(pool_size=(2, 2)))

# Flatten Layer (1차원으로 변환)
model.add(Flatten())

# 완전 연결층 (Fully Connected Layer)
model.add(Dense(128, activation='relu'))

# 출력층 (Output Layer)
model.add(Dense(10, activation='softmax'))

# 모델 컴파일
model.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])

# 모델 요약
model.summary()
```

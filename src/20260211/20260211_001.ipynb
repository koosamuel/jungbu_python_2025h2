## 주문 분석
* customer_master.csv       : 고객데이터, 이름, 성별 등

* transaction_1.csv         : 구매내역 데이터

```python
customer_master = pd.read_csv('../../data/customer_master.csv')
customer_master.head()
```

```python
item_master = pd.read_csv('../../data/item_master.csv')
item_master.head()
```

```python
transaction_1 = pd.read_csv('../../data/transaction_1.csv')
transaction_1.head()
```

```python
transaction_2 = pd.read_csv('../../data/transaction_2.csv')
transaction_2.head()
```

```python
transaction_detail_1 = pd.read_csv('../../data/transaction_detail_1.csv')
transaction_detail_1.head()
```

```python
transaction_detail_2 = pd.read_csv('../../data/transaction_detail_2.csv')
transaction_detail_2.head()
```

```python
transaction = pd.concat([transaction_1, transaction_2])   ## 이어붙이기
transaction
```

```python
transaction_detail = pd.concat([transaction_detail_1,transaction_detail_2]) ## 이어 붙이기
transaction_detail
```

```python
## 이어 붙인 자료를 csv 파일로 저장

transaction.to_csv('../../data/transaction.csv')
transaction_detail.to_csv('../../data/transaction_detail.csv')
```

```python
##  매출 데이터끼리 merge

join_data = pd.merge(transaction_detail, 
                      transaction[["transaction_id", "payment_date", "customer_id"]], 
                      on="transaction_id", how="left")
join_data.head()
```

```python
print(len(transaction_detail))
print(len(transaction))
print(len(join_data))
join_data_customer = pd.merge(join_data, 
                              customer_master, 
                              on='customer_id', 
                              how='left')
join_data_customer
```

```python
graph_data = pd.pivot_table(join_data_item, 
                            index='payment_month', 
                            columns='item_name', 
                            values='price', 
                            aggfunc='sum')
graph_data
```

```python
plt.plot(list(graph_data.index), graph_data["PC-A"], label='PC-A')
plt.plot(list(graph_data.index), graph_data["PC-B"], label='PC-B')
plt.plot(list(graph_data.index), graph_data["PC-C"], label='PC-C')
plt.plot(list(graph_data.index), graph_data["PC-D"], label='PC-D')
plt.plot(list(graph_data.index), graph_data["PC-E"], label='PC-E')
plt.legend()  
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.lineplot(data=graph_data, markers=True, dashes=False) # markers: 점 표시, dashes: 실선 유지

plt.title("Monthly PC Sales")
plt.xlabel("Payment Month")
plt.ylabel("Sales")
plt.legend(title="Item Name")
plt.show()
```
